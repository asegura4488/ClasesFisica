```python
import sympy as sp

# --- 1. Pauli y operadores de un único espín ----------------------------
sigma_x = sp.Matrix([[0, 1], [1, 0]])
sigma_y = sp.Matrix([[0, -sp.I], [sp.I, 0]])
sigma_z = sp.Matrix([[1, 0], [0, -1]])

Jx_1 = sp.Rational(1, 2) * sigma_x
Jy_1 = sp.Rational(1, 2) * sigma_y
Jz_1 = sp.Rational(1, 2) * sigma_z
I2   = sp.eye(2)

# --- 2. Operadores totales en la base desacoplada -----------------------
Jx = sp.kronecker_product(Jx_1, I2) + sp.kronecker_product(I2, Jx_1)
Jy = sp.kronecker_product(Jy_1, I2) + sp.kronecker_product(I2, Jy_1)
Jz = sp.kronecker_product(Jz_1, I2) + sp.kronecker_product(I2, Jz_1)

# --- 3. Matriz de cambio de base (Clebsch–Gordan) -----------------------
sqrt2 = sp.sqrt(2)
U = sp.Matrix([
    [1, 0, 0, 0],
    [0, 1/sqrt2, 1/sqrt2, 0],
    [0, 0, 0, 1],
    [0, 1/sqrt2, -1/sqrt2, 0]
])

# --- 4. Operadores en la base acoplada ----------------------------------
Jx_ac = (U * Jx * U.T).applyfunc(sp.simplify)
Jy_ac = (U * Jy * U.T).applyfunc(sp.simplify)
Jz_ac = (U * Jz * U.T).applyfunc(sp.simplify)
J2    = Jx**2 + Jy**2 + Jz**2
J2_ac = (U * J2 * U.T).applyfunc(sp.simplify)

# --- 5. Mostrar en pantalla o exportar a LaTeX --------------------------
print("Jx desacoplada:\n", sp.pretty(Jx))
print("\nJx acoplada:\n", sp.pretty(Jx_ac))
print("\nJ² acoplada:\n", sp.pretty(J2_ac))

# Si prefieres LaTeX:
print("\nLaTeX Jx desacoplada:\n", sp.latex(Jx))
print("\nLaTeX Jx acoplada:\n", sp.latex(Jx_ac))
print("\nLaTeX J² acoplada:\n", sp.latex(J2_ac))
```
