```python
# Momento angular 1
import sympy as sp
h = sp.symbols('hbar')

# operadores de subida/bajada
Splus  = h*sp.sqrt(2)*sp.Matrix([[0,1,0],[0,0,1],[0,0,0]])
Sminus = Splus.T.conjugate()

Sx = (Splus + Sminus)/2
Sy = (Splus - Sminus)/(2*sp.I)
Sz = h*sp.Matrix([[1,0,0],[0,0,0],[0,0,-1]])

S2 = Sx**2 + Sy**2 + Sz**2
```
