```python
import sympy as sp
h = sp.symbols('hbar')

# --- Espín-1 ----------------------------------------------------------
sx1 = (h/sp.sqrt(2))*sp.Matrix([[0,1,0],
                                [1,0,1],
                                [0,1,0]])
sy1 = (h/sp.sqrt(2))*sp.Matrix([[0,-sp.I,0],
                                [sp.I,0,-sp.I],
                                [0,sp.I,0]])
sz1 =  h*sp.Matrix([[1,0,0],
                    [0,0,0],
                    [0,0,-1]])

# --- Espín-½ ----------------------------------------------------------
sx2 = (h/2)*sp.Matrix([[0,1],[1,0]])
sy2 = (h/2)*sp.Matrix([[0,-sp.I],[sp.I,0]])
sz2 = (h/2)*sp.Matrix([[1,0],[0,-1]])

# --- Operadores totales ----------------------------------------------
kp = sp.kronecker_product
Sx = kp(sx1, sp.eye(2)) + kp(sp.eye(3), sx2)
Sy = kp(sy1, sp.eye(2)) + kp(sp.eye(3), sy2)
Sz = kp(sz1, sp.eye(2)) + kp(sp.eye(3), sz2)
S2 = sp.simplify(Sx**2 + Sy**2 + Sz**2)

# --- Autovectores de S²  ---------------------------------------------
def normalise(v):
    v = sp.Matrix(v)
    n = sp.sqrt((v.H*v)[0])
    return v / n

states = []                      # [(λ_S2, m, vector)]
for ev, mult, vecs in S2.eigenvects():
    for v in vecs:
        v  = normalise(v)
        m  = sp.simplify((v.H*Sz*v)[0] / (h*(v.H*v)[0]))   # eigen-m
        states.append((ev, m, v))

# Orden: J = 3/2 (λ = 15/4 ℏ²) primero, m ↓; luego J = 1/2
states.sort(key=lambda t: (-float(t[0].subs(h,1)), -float(t[1])))

# Matriz de Clebsch–Gordan U
U = sp.Matrix.hstack(*[v for _,_,v in states])


S2
```
