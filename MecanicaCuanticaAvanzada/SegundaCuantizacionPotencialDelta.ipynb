```python
import numpy as np
import matplotlib.pyplot as plt

# Parámetros físicos
U0 = 5.0         # intensidad del potencial
r0 = 2.0         # posición del delta
N  = 50         # número de puntos en k

# Rango de k (simétrico)
k = np.linspace(-4, 4, N)

# Construcción de la matriz H_{ij} = U0 * exp(-i (k_i - k_j) r0)
Kdiff = np.subtract.outer(k, k)   # matriz de diferencias (k_i - k_j)
H = U0 * np.exp(-1j * Kdiff * r0) # Hamiltoniano complejo

# --- Magnitud y fase ---
Habs  = np.abs(H)
Hphase = np.angle(H)   # fase en radianes (-pi, pi)

# --- Figura ---
fig, axs = plt.subplots(1, 3, figsize=(15,5))

# 1️⃣ Magnitud
im1 = axs[0].imshow(Habs, cmap='viridis', origin='lower', extent=[k[0],k[-1],k[0],k[-1]])
axs[0].set_title('|H(k,k\')|')
axs[0].set_xlabel("k'")
axs[0].set_ylabel("k")
fig.colorbar(im1, ax=axs[0])

# 2️⃣ Fase (en radianes)
im2 = axs[1].imshow(Hphase, cmap='twilight', origin='lower', extent=[k[0],k[-1],k[0],k[-1]])
axs[1].set_title('Fase[H(k,k\')] (radianes)')
axs[1].set_xlabel("k'")
axs[1].set_ylabel("k")
fig.colorbar(im2, ax=axs[1])

# 3️⃣ Fase como círculo de color (HSV)
# Convertimos magnitud y fase a un color HSV → RGB
HSV = np.zeros((N, N, 3))
HSV[...,0] = (Hphase + np.pi) / (2*np.pi)  # tono = fase normalizada
HSV[...,1] = 1.0                           # saturación total
HSV[...,2] = Habs / np.max(Habs)           # brillo = magnitud normalizada
RGB = plt.cm.hsv(HSV[...,0]) * HSV[...,2][...,None]
axs[2].imshow(RGB, origin='lower', extent=[k[0],k[-1],k[0],k[-1]])
axs[2].set_title('Mapa complejo H(k,k\')')
axs[2].set_xlabel("k'")
axs[2].set_ylabel("k")

plt.tight_layout()
plt.show()
import numpy as np
import matplotlib.pyplot as plt

# Parámetros físicos
U0 = 5.0         # intensidad del potencial
r0 = 1.0         # posición del delta
N  = 20         # número de puntos en k

# Rango de k (simétrico)
k = np.linspace(-4, 4, N)

# Construcción de la matriz H_{ij} = U0 * exp(-i (k_i - k_j) r0)
Kdiff = np.subtract.outer(k, k)   # matriz de diferencias (k_i - k_j)
H = U0 * np.exp(-1j * Kdiff * r0) # Hamiltoniano complejo

# --- Magnitud y fase ---
Habs  = np.abs(H)
Hphase = np.angle(H)   # fase en radianes (-pi, pi)

# --- Figura ---
fig, axs = plt.subplots(1, 3, figsize=(15,5))

# 1️⃣ Magnitud
im1 = axs[0].imshow(Habs, cmap='viridis', origin='lower', extent=[k[0],k[-1],k[0],k[-1]])
axs[0].set_title('|H(k,k\')|')
axs[0].set_xlabel("k'")
axs[0].set_ylabel("k")
fig.colorbar(im1, ax=axs[0])

# 2️⃣ Fase (en radianes)
im2 = axs[1].imshow(Hphase, cmap='twilight', origin='lower', extent=[k[0],k[-1],k[0],k[-1]])
axs[1].set_title('Fase[H(k,k\')] (radianes)')
axs[1].set_xlabel("k'")
axs[1].set_ylabel("k")
fig.colorbar(im2, ax=axs[1])

# 3️⃣ Fase como círculo de color (HSV)
# Convertimos magnitud y fase a un color HSV → RGB
HSV = np.zeros((N, N, 3))
HSV[...,0] = (Hphase + np.pi) / (2*np.pi)  # tono = fase normalizada
HSV[...,1] = 1.0                           # saturación total
HSV[...,2] = Habs / np.max(Habs)           # brillo = magnitud normalizada
RGB = plt.cm.hsv(HSV[...,0]) * HSV[...,2][...,None]
axs[2].imshow(RGB, origin='lower', extent=[k[0],k[-1],k[0],k[-1]])
axs[2].set_title('Mapa complejo H(k,k\')')
axs[2].set_xlabel("k'")
axs[2].set_ylabel("k")

plt.tight_layout()
plt.show()
```
