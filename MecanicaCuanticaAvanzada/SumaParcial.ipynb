```python
import sympy as sp

# --- Espín-1 ----------------------------------------------------------
sx1 = (1/sp.sqrt(2))*sp.Matrix([[0,1,0],
                                [1,0,1],
                                [0,1,0]])
sy1 = (1/sp.sqrt(2))*sp.Matrix([[0,-sp.I,0],
                                [sp.I,0,-sp.I],
                                [0,sp.I,0]])
sz1 =  1*sp.Matrix([[1,0,0],
                    [0,0,0],
                    [0,0,-1]])

# --- Espín-½ ----------------------------------------------------------
sx2 = (1/2)*sp.Matrix([[0,1],[1,0]])
sy2 = (1/2)*sp.Matrix([[0,-sp.I],[sp.I,0]])
sz2 = (1/2)*sp.Matrix([[1,0],[0,-1]])
```

```python
kp = sp.kronecker_product
```

```python
I1 = kp(sx1,sx2)
I1
```

```python
I2 = kp(sy1,sy2)
I2
```

```python
I3 = kp(sz1,sz2)
I3
```

```python
K = I1+I2+I3
K
```

```python
# Veamos el otro termino
J12 = sx1**2+sy1**2+sz1**2
J12
```

```python
kp(J12,sp.eye(2))
```

```python
J22 = sx2**2+sy2**2+sz2**2
J22
```

```python
kp(sp.eye(3),J22)
```

```python
J2 = 11/4*kp(sp.eye(3),sp.eye(2)) + 2*K
J2
```
