# Movimiento en un sistema rotante y en uno inercial

La partícula se mueve con velocidad constante respecto al sistema rotante: $\mathbf r'(t)=vt(\cos\theta,\sin\theta)$. Desde el laboratorio describe una espiral: $\mathbf r(t)=vt(\cos(\theta+\omega t),\sin(\theta+\omega t))$. Las flechas rojas indican **velocidades** (se muestran con el mismo factor de escala visual en ambos paneles).

Ejecuta la celda siguiente en Google Colab.

```python
import numpy as np
import matplotlib.pyplot as plt
from matplotlib.animation import FuncAnimation
from IPython.display import HTML, display

# Parámetros
v = 1.0
omega = 0.8            # rad/s
theta = np.deg2rad(30)
t = np.linspace(0, 6, 90)

# Sistema rotante: trayectoria recta y velocidad constante
xp = v * t * np.cos(theta)
yp = v * t * np.sin(theta)
vxp = v * np.cos(theta) * np.ones_like(t)
vyp = v * np.sin(theta) * np.ones_like(t)

# Sistema inercial: trayectoria espiral y velocidad creciente
phi = theta + omega * t
x = v * t * np.cos(phi)
y = v * t * np.sin(phi)
vx = v * np.cos(phi) - omega * v * t * np.sin(phi)
vy = v * np.sin(phi) + omega * v * t * np.cos(phi)

fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(11, 5))
for ax, title in [(ax1, "Sistema rotante S′"), (ax2, "Sistema inercial S")]:
    ax.set(xlim=(-7, 7), ylim=(-7, 7), xlabel="x", ylabel="y", title=title)
    ax.set_aspect("equal")
    ax.grid(alpha=0.25)

line1, = ax1.plot([], [], color="tab:blue", lw=2)
line2, = ax2.plot([], [], color="tab:blue", lw=2)
point1, = ax1.plot([], [], "o", color="tab:blue", ms=8)
point2, = ax2.plot([], [], "o", color="tab:blue", ms=8)

# Flechas rojas: velocidades (misma escala visual en ambos paneles)
arrow_scale = 0.38
arrow1 = ax1.quiver([0], [0], [0], [0], color="tab:red", angles="xy", scale_units="xy", scale=1, width=0.012)
arrow2 = ax2.quiver([0], [0], [0], [0], color="tab:red", angles="xy", scale_units="xy", scale=1, width=0.012)
time_label = fig.suptitle("")

def update(i):
    line1.set_data(xp[:i+1], yp[:i+1])
    line2.set_data(x[:i+1], y[:i+1])
    point1.set_data([xp[i]], [yp[i]])
    point2.set_data([x[i]], [y[i]])
    arrow1.set_offsets([[xp[i], yp[i]]])
    arrow2.set_offsets([[x[i], y[i]]])
    arrow1.set_UVC(arrow_scale * vxp[i], arrow_scale * vyp[i])
    arrow2.set_UVC(arrow_scale * vx[i], arrow_scale * vy[i])
    time_label.set_text(f"t = {t[i]:.1f} s     |     v′ = {v:.1f}     |     |v laboratorio| = {np.hypot(vx[i], vy[i]):.2f}")
    return line1, line2, point1, point2, arrow1, arrow2, time_label

anim = FuncAnimation(fig, update, frames=len(t), interval=70, blit=False)
plt.close(fig)   # Evita una figura estática adicional en Colab
display(HTML(anim.to_jshtml()))
```
