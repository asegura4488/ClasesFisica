```python
import numpy as np
import matplotlib.pyplot as plt


# Eventos: (x, t)
E1 = (2, 3)
E2 = (-1, 4)
E3 = (-1, -2)


# Límites del gráfico
x_min, x_max = -6, 6
t_min, t_max = -6, 6

# ============================================================
# CÁLCULOS
# ============================================================

x = np.linspace(x_min, x_max, 500)
t = np.linspace(t_min, t_max, 500)

# Cono de luz de E1
t_future = E1[1] + np.abs(x - E1[0])
t_past   = E1[1] - np.abs(x - E1[0])


# ============================================================
# GRÁFICA
# ============================================================

plt.figure(figsize=(7,7))

# Cono de luz
plt.plot(x, t_future, 'k--', label="Cono de luz futuro (E1)")
plt.plot(x, t_past,   'k--', label="Cono de luz pasado (E1)")

# Eventos
plt.scatter(*E1, color='blue', zorder=5)
plt.text(E1[0]+0.1, E1[1]+0.1, f"E1 {E1}")

plt.scatter(*E2, color='red', zorder=5)
plt.text(E2[0]+0.1, E2[1]+0.1, f"E2 {E2}")

plt.scatter(*E3, color='green', zorder=5)
plt.text(E3[0]+0.1, E3[1]+0.1, f"E3 {E3}")


# Ejes del sistema S
plt.axhline(0, color='gray', linewidth=0.8)
plt.axvline(0, color='gray', linewidth=0.8)

plt.xlabel("x")
plt.ylabel("t")
plt.title("Diagrama espacio–tiempo")
plt.legend()
plt.grid(True)

plt.xlim(x_min, x_max)
plt.ylim(t_min, t_max)

plt.show()
```

```python
# Velocidad del sistema S', para le punto a

v = -1/3   # para que E1 y E2 sean simultáneos

# Eventos (x, t)
E1 = (2, 3)
E2 = (-1, 4)

# Rango
t = np.linspace(-6, 6, 400)
x = np.linspace(-6, 6, 400)

# Ejes primados
# eje t' : x = v t
x_tprime = v * t

# eje x' : t = v x
t_xprime = v * x

plt.figure(figsize=(6,6))

# Ejes del sistema S
plt.axhline(0, color='black')
plt.axvline(0, color='black')

# Ejes del sistema S'
plt.plot(x_tprime, t, color='blue', linewidth=2, label="eje t'  (x'=0)")
plt.plot(x, t_xprime, color='purple', linewidth=2, label="eje x' (t'=0)")

# Eventos
plt.scatter(E1[0], E1[1], color='red', zorder=5)
plt.text(E1[0]+0.1, E1[1]+0.1, f"E1 {E1}", fontsize=12)

plt.scatter(E2[0], E2[1], color='green', zorder=5)
plt.text(E2[0]+0.1, E2[1]+0.1, f"E2 {E2}", fontsize=12)


# aca ponemos que los eventos efectivamente ocurren al mismo tiempo
# Línea de simultaneidad t' = const que pasa por E1
# t = v (x - x1) + t1

x_line = np.linspace(-6, 6, 400)
t_line = v * (x_line - E1[0]) + E1[1]

plt.plot(x_line,t_line,color='black',linestyle='--',linewidth=2,label="t' = const (E1 y E2)")


plt.xlabel("x")
plt.ylabel("t")
plt.legend()
plt.grid()
plt.xlim(-6,6)
plt.ylim(-6,6)

plt.show()
```

```python
# Velocidad del sistema S' (inciso c)
v = 2/3

# Eventos (z, t)
E2 = (2, 4)
E3 = (-2, -2)

# Rango
t = np.linspace(-6, 6, 400)
z = np.linspace(-6, 6, 400)

# Ejes primados
# eje t' : z = v t
z_tprime = v * t

# eje z' : t = v z
t_zprime = v * z

plt.figure(figsize=(6,6))

# Ejes del sistema S
plt.axhline(0, color='black')
plt.axvline(0, color='black')

# Ejes del sistema S'
plt.plot(z_tprime, t, color='blue', linewidth=2, label="eje t'  (z'=0)")
plt.plot(z, t_zprime, color='purple', linewidth=2, label="eje z' (t'=0)")

# Eventos

plt.scatter(E2[0], E2[1], color='green', zorder=5)
plt.text(E2[0]+0.1, E2[1]+0.1, f"E2 {E2}", fontsize=12)

plt.scatter(E3[0], E3[1], color='red', zorder=5)
plt.text(E3[0]+0.1, E3[1]+0.1, f"E3 {E3}", fontsize=12)

# Línea z' = const que pasa por E2 (y E3)
z_line = v * (t - E2[1]) + E2[0]
plt.plot(z_line, t, color='black', linestyle='--',
         linewidth=2, label="z' = const (E2 y E3)")

plt.xlabel("z")
plt.ylabel("t")
plt.legend()
plt.grid()
plt.xlim(-6,6)
plt.ylim(-6,6)

plt.savefig("diagrama_minkowski.pdf", dpi=300, bbox_inches="tight")
plt.show()
```
