```python
import numpy as np
import matplotlib.pyplot as plt
from google.colab import files

alpha = 1/137.035999084
m_mu = 0.1056583755   # GeV

# conversion: 1 GeV^{-2} = 3.89379e5 nb
GEV2_TO_NB = 3.89379e5

def beta_mu(s):
    return np.sqrt(1 - 4*m_mu**2/s)

def sigma_total(s):
    beta = beta_mu(s)
    return (2*np.pi*alpha**2/(3*s)) * beta * (3 - beta**2)

sqrt_s = np.linspace(2*m_mu + 1e-4, 1.5, 600)
s_vals = sqrt_s**2
sigma_nb = sigma_total(s_vals) * GEV2_TO_NB

plt.figure(figsize=(8,5))
plt.plot(sqrt_s, sigma_nb, lw=2)
plt.xlabel(r'$\sqrt{s}\ \mathrm{[GeV]}$')
plt.ylabel(r'$\sigma_{\mathrm{tot}}\ \mathrm{[nb]}$')
plt.title(r'Sección transversal total para $e^-e^+\to \mu^-\mu^+$')
plt.grid(True)
plt.tight_layout()
plt.savefig("SeccionTotal.pdf", dpi=200)
plt.show()
files.download("SeccionTotal.pdf")
```
