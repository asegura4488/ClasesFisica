```python
import numpy as np
import matplotlib.pyplot as plt

# ==========================================================
# 1) Función de Källén — masas hijas idénticas
# ==========================================================

# Variable adimensional r = m_chi / m1
r = np.linspace(0, 0.5, 500)

# λ(m1^2, m_chi^2, m_chi^2)/m1^4
# Para masas idénticas: λ/m1^4 = 1 - 4 r^2
kallen_identical = 1 - 4 * r**2

plt.figure()

plt.plot(r, kallen_identical)
plt.xlim(0, 0.5)
plt.ylim(0, 1)

plt.xlabel(r"$m_\chi/m_1$")
plt.ylabel(r"$\lambda(m_1^2,m_\chi^2,m_\chi^2)/m_1^4$")
plt.title("Función de Källén para masas hijas idénticas")

plt.tight_layout()
plt.savefig("kallen.pdf")
plt.show()


# ==========================================================
# 2) Mapa de calor — masas hijas diferentes
# ==========================================================

# Variables adimensionales
r2 = np.linspace(0, 1, 500)
r3 = np.linspace(0, 1, 500)

R2, R3 = np.meshgrid(r2, r3)

# λ(1, r2^2, r3^2)
kallen_map = (
    1
    + R2**4
    + R3**4
    - 2 * R2**2
    - 2 * R3**2
    - 2 * R2**2 * R3**2
)

# Solo región físicamente permitida (λ ≥ 0)
kallen_positive = np.maximum(kallen_map, 0)

plt.figure()

plt.imshow(
    kallen_positive,
    extent=[0, 1, 0, 1],
    origin="lower",
    aspect="equal",
    vmin=0,
    vmax=1
)

plt.colorbar(label=r"$\lambda/m_1^4$")

# Recta r2 + r3 = 1
line = np.linspace(0, 1, 500)
plt.plot(line, 1 - line)

plt.xlim(0, 1)
plt.ylim(0, 1)

plt.xlabel(r"$m_2/m_1$")
plt.ylabel(r"$m_3/m_1$")
plt.title(r"Mapa de $\lambda(m_1^2,m_2^2,m_3^2)/m_1^4$")

plt.tight_layout()
plt.savefig("MapaKallen.pdf")
plt.show()
```
